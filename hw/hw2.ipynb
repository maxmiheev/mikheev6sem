````python

#t = np.range(-2*pi, 2*pi, 0.1)
#y = np.sin(t)
#plt.plot(t,y,'b--')
#plt.show()

x = np.arange(-2*math.pi, 2*math.pi, 0.1)

def func(x):
   return 2*np.sin(x)-np.cos(2*x)
def dfunc(x):
  return 2*np.cos(x)+2*np.sin(2*x)

#plt.plot(x,func(x),'b-', label="f(x)=2sinx-cos2x")
#plt.plot(x, dfunc(x), 'r--', label="f'(x)=2cosx+2sin2x")

def beauty_derivative_plot(t):
  z = np.linspace(t-1, t+1, 100)
  y = dfunc(t)*(z-t)+func(t)

  plt.plot(z, y, 'g-', label=f"tanget at x={t}")
  plt.plot(x,func(x),'b-', label="f(x)=2sinx-cos2x")
  plt.plot(x, dfunc(x), 'r--', label="f'(x)=2cosx+2sin2x")

  plt.annotate(f"f'({t})={dfunc(t):.2f}", xy=(t, func(t)), xytext=(t-2,func(t)+2),
                           color="black", fontsize=10,
                           arrowprops={"facecolor": "lightgreen",
                                       "width": 0.5,
                                       "headwidth": 3,
                                       "headlength": 3})
  plt.grid(True)
  plt.legend(loc="best")
  plt.show()

beauty_derivative_plot(-3)

#plt.grid(True)
#plt.legend(loc="best")
#plt.show()

beauty_derivative_plot(-2)
#plt.grid(True)
#plt.legend(loc="best")
#plt.show()

beauty_derivative_plot(1.7)
#plt.grid(True)
#plt.legend(loc="best")
#plt.show()

````

- Plot the population size of each species for each year on a single graph. Add legend and axis labels to the plot.

````python

years = data[:, 0].astype(int)
hares = data[:, 1]
lynxes = data[:, 2]
carrots = data[:, 3]
plt.figure(figsize=(10, 6))
plt.plot(years, hares, 'b-', label='Hares')
plt.plot(years, lynxes, 'r-', label='Lynxes')
plt.plot(years, carrots, 'g-', label='Carrots')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Population Dynamics')
plt.legend()
plt.grid(True)
plt.show()

````

- Find which year each species had the largest population. Mark these years in the plot of populations.

````python
max_hares = np.argmax(hares)
max_lynxes = np.argmax(lynxes)
max_carrots = np.argmax(carrots)

````

````python

plt.plot(years, hares, 'b-', label='Hares')
plt.plot(years, lynxes, 'r-', label='Lynxes')
plt.plot(years, carrots, 'g-', label='Carrots')
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Year with largest population')

plt.scatter(years[max_hares], hares[max_hares], s=150)
plt.scatter(years[max_lynxes], lynxes[max_lynxes], s=150)
plt.scatter(years[max_carrots], carrots[max_carrots], s=150)


plt.legend()
plt.grid(True)
plt.show()
````

- Assuming total population of hares and lynxes is 100%, plot stacked bar graph showing ratio of each specis for each year in the period.

````python
total_population = hares + lynxes

h_ratio = (hares / total_population) * 100
l_ratio = (lynxes / total_population) * 100

plt.figure(figsize=(10, 6))
plt.plot(years, h_ratio, 'b-', label='Ratio of hares')
plt.plot(years, l_ratio, 'r-', label='ratio of lynxes')
plt.xlabel('Year')
plt.ylabel('Ratio')
plt.title('Ratio')

plt.legend()
plt.grid(True)
plt.show()
````

- Find the mean and std of the populations of each species. Plot the historgam of population for each species and show mean values with vertical line. Show each histogram in a separate plot. Arrange 3 plots in a row.

````python
mean_hares = np.mean(hares)
mean_lynxes = np.mean(lynxes)
mean_carrots = np.mean(carrots)

std_hares = np.std(hares)
std_lynxes = np.std(lynxes)
std_carrots = np.std(carrots)

````

````python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))

axes[0].hist(hares)
axes[0].axvline(mean_hares, label = f'Mean = {mean_hares:.2f}; Std = {std_hares:.2f}')
axes[0].set_title("Hares")
axes[0].set_xlabel("Population")
axes[0].set_ylabel("Frequency")
axes[0].legend()

axes[1].hist(lynxes)
axes[1].axvline(mean_lynxes, label = f'Mean = {mean_lynxes:.2f}; Std = {std_lynxes:.2f}')
axes[1].set_title("Lynxes")
axes[1].set_xlabel("Population")
axes[1].set_ylabel("Frequency")
axes[1].legend()

axes[2].hist(carrots)
axes[2].axvline(mean_carrots, label = f'Mean = {mean_carrots:.2f}; Std = {std_carrots:.2f}')
axes[2].set_title("Carrots")
axes[2].set_xlabel("Population")
axes[2].set_ylabel("Frequency")
axes[2].legend()




plt.show()
````

- Find which species (hares or lynxes) has the largest population for each year. Print the result as [H, H, L, H, ...]. Plot a pie chart showing the ratio of "H" and "L" values.

````python
result = np.where(hares > lynxes, "H", "L")

print(result)
````

````python
sum_hares = np.sum(result == "H")
sum_lynxes = np.sum(result == "L")

plt.figure(figsize=(5,5))
plt.pie([sum_hares, sum_lynxes], labels = ["Hares", "Lynxes"])
plt.title("Ratio H vs L")
plt.show()
````

- Plot the change in the hare population and in the lynx population for each year. Find the correlation coefficient between the both time series.

````python
change_hares = np.diff(hares)
change_lynxes = np.diff(lynxes)

change_years = years[1:]

plt.plot(change_years, change_hares, label = "Change of the hare's population")
plt.plot(change_years, change_lynxes, label = "Change of the lynx's population")


plt.xlabel("Year")
plt.ylabel("Population change")
plt.title("Change in the hare population and in the lynx population for each year")
plt.legend()
plt.grid(True)
plt.show()
````

````python
correlation_coef = np.corrcoef(change_hares, change_lynxes)
print(f'Correlation = {correlation_coef[1,0]}')
````
